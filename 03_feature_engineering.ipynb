```python
import pandas as pd

df_model = pd.read_csv("../data/processed/churn_clean_model.csv")
df_model.head()
```

```python
df_model["HighSupportCalls"] = (df_model["Support Calls"] > 5).astype(int)
```

```python
df_model["DelayRatio"] = df_model["Payment Delay"] / (df_model["Tenure"] + 1)
```

```python
df_model["SpendPerMonth"] = df_model["Total Spend"] / (df_model["Tenure"] + 1)
```

```python
df_model["PremiumHighUsage"] = (
    (df_model["Subscription Type_Premium"] == 1) & (df_model["Usage Frequency"] > 20)
).astype(int)
```

```python
df_model = df_model.drop(columns=["CustomerID"])
```

```python
print(df_model.columns)
```

```python
df_model = df_model.drop(columns=["CustomerID"], errors="ignore")
```

```python
df_model.to_csv("../data/processed/churn_features.csv", index=False)
```
