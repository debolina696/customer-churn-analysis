```python
## Churn Drivers
- **Payment Delay** → Customers with frequent delays are highly likely to churn.
- **Support Calls** → High support calls signal dissatisfaction and increase churn risk.
- **Tenure** → Longer tenure reduces churn probability.
- **SpendPerMonth** → Higher monthly spend correlates with lower churn.
- **Usage Frequency** → Active users are less likely to churn.
```

```python
## Model Comparison
| Model               | Accuracy | Precision | Recall | F1 Score | AUC |
|----------------------|----------|-----------|--------|----------|-----|
| Logistic Regression  | 0.83     | 0.81      | 0.84   | 0.83     | 0.91 |
| Random Forest        | 0.99     | 0.99      | 0.99   | 0.99     | 1.00 |

- Logistic Regression → Good baseline, interpretable.
- Random Forest → Best performing model, near-perfect accuracy.
```

```python
## Recommendations
- **Target customers with frequent payment delays** → Offer flexible payment plans or reminders.
- **Monitor high-support-call customers** → Provide proactive support to reduce dissatisfaction.
- **Reward long-tenure customers** → Loyalty programs to strengthen retention.
- **Encourage higher usage frequency** → Engagement campaigns to keep customers active.
```

```python
## Conclusion
The analysis shows that churn is strongly driven by payment behavior and customer support interactions. 
Random Forest is the recommended model for deployment due to its superior predictive power. 
Logistic Regression remains useful for explainability in business discussions.
```
