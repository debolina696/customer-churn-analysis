```python
import os
import pandas as pd

# List all files in raw folder
raw_path = "../data/raw"
files = os.listdir(raw_path)
print("Files in raw folder:", files)

# Pick the first CSV file automatically
for f in files:
    if f.endswith(".csv"):
        file_path = os.path.join(raw_path, f)
        break

# Load dataset with safe encoding
df = pd.read_csv(file_path, encoding="ISO-8859-1")
print("Loaded file:", file_path)
print(df.shape)
df.head()
```

```python
df.info()
df.describe(include='all')
```

```python
# Drop rows with any missing values (or impute if needed)
df = df.dropna()

# Alternative: fill numeric with median, categorical with mode
for col in df.select_dtypes(include=['float64','int64']).columns:
    df[col] = df[col].fillna(df[col].median())

for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
```

```python
from sklearn.preprocessing import StandardScaler

num_cols = df_encoded.select_dtypes(include=['float64','int64']).columns
scaler = StandardScaler()
df_encoded[num_cols] = scaler.fit_transform(df_encoded[num_cols])
```

```python
df_encoded.to_csv("../data/processed/churn_clean.csv", index=False)
```
